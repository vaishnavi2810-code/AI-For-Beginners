## Evaluating Retrieval

To introduce the problem, we take some example Chinese paragraphs to use as queries and English paragraphs to use as candidate results to search through.

```python
query_articles = [r['text'] for r in articles if r['lang'] == 'zh']
result_articles = [r['text'] for r in articles if r['lang'] == 'en']
```

To make the example clearer, we will use different numbers of queries and results.

```python
qembed = labse.encode(query_articles[0:200])
rembed = labse.encode(result_articles[0:500])
```

Multiplying the query embeddings by the result embeddings, we get a $200 \times 500$ queries-by-results matrix.

```python
sim = qembed @ rembed.T
sim.shape
```

We use numpy's `argmax` function along the second dimension (`axis=1`) to get the index of the top result for each query.

```python
argmax = np.argmax(sim, axis=1)
argmax
```

Since the query and result documents are in the same order, matching Chinese and English documents have the same index. This allows us to compute the accuracy, or &ldquo;recall at 1&rdquo;, of Chinese-to-English retrieval.

```python
sum([a==b for (a, b) in zip(range(len(argmax)), argmax)])/len(argmax)
```

Your first task is to compute the recall at 1 for Arabic, Chinese, French, German, and Greek query documents matching English documents. Use the first 1000 English documents as the candidates you will search through.

```python
candidates = labse.encode(result_articles[0:1000])
```

For each of the other five languages, construct embeddings for the first 1000 documents and measure how often the most similar English document is the matching one.

```python
# TODO: Compute and print the recall at 1 for X-English retrieval
# where X \in {ar,de,el,fr,zh}

from sklearn.metrics.pairwise import cosine_similarity
langs = ['ar', 'de', 'el', 'fr', 'zh']
candidates = labse.encode(result_articles[0:1000])
candidate_articles = [r for r in articles if r['lang'] == 'en'][0:1000]

for lang in langs:
  query_articles = [r for r in articles if r['lang'] == lang]
  queries = labse.encode([r['text'] for r in query_articles])
  similarities = cosine_similarity(queries, candidates)
  argmax = np.argmax(similarities, axis=1)
  correct_count = 0
  total_with_match = 0
  for i, query_article in enumerate(query_articles):
    query_id = query_article['id']
    for j, en_article in enumerate(candidate_articles):
      if en_article['id'] == query_id:
        total_with_match += 1
        if argmax[i] == j:
          correct_count += 1
        break
  recall_at_1 = correct_count / total_with_match if total_with_match > 0 else 0
  print(f"{lang} → en: {recall_at_1:.3f} ({recall_at_1*100:.1f}%) "
        f"[{correct_count}/{total_with_match} correct]")
```

We often use retrieved documents to provide extra context to a language model. In that case, we might retrieve more than one document per query to increase the likelihood that useful documents are in the top $k$. For each of the five non-English languages, write code to evaluate the **recall at k** (R@k), i.e., the proportion of queries for which the correct document was anywhere in the top k results.

```python
# TODO: Write a function to compute recall at k
def compute_recall_at_k(similarities, query_articles, candidate_articles, k):
  top_k_indices = np.argsort(similarities, axis=1)[:, ::-1][:, :k]
  correct_count = 0
  total_with_match = 0
  for i, query_article in enumerate(query_articles):
    query_id = query_article['id']
    for j, en_article in enumerate(candidate_articles):
      if en_article['id'] == query_id:
        total_with_match += 1
        if j in top_k_indices[i]:
          correct_count += 1
        break
  recall_at_k = correct_count / total_with_match if total_with_match > 0 else 0
  return recall_at_k, correct_count, total_with_match
```

```python
# TODO: Compute and print recall at 5 and recall at 10 for X-English retrieval
# where X \in {ar,de,el,fr,zh}
langs = ['ar', 'de', 'el', 'fr', 'zh']
candidates = labse.encode(result_articles[0:1000])
candidate_articles = [r for r in articles if r['lang'] == 'en'][0:1000]

print("Recall@k for Cross-Language Retrieval (X → English):")
results = {lang: {} for lang in langs}

for lang in langs:
  query_articles = [r for r in articles if r['lang'] == lang]
  queries = labse.encode([r['text'] for r in query_articles])
  similarities = cosine_similarity(queries, candidates)
  k_values = [1, 5, 10]
  print(f"\n{lang.upper()} → EN:")
  print("-" * 70)
  for k in k_values:
    recall, correct, total = compute_recall_at_k(
        similarities, query_articles, candidate_articles, k=k
    )
    results[lang][k] = recall
    print(f"  Recall@{k:2d}: {recall:.3f} ({recall*100:.1f}%) "
          f"[{correct}/{total} correct]")

print("\n" + "=" * 70)
print("SUMMARY TABLE:")
print("=" * 70)
print(f"{'Language':<12} {'Recall@1':<12} {'Recall@5':<12} {'Recall@10':<12}")
print("-" * 70)

for lang in langs:
  print(f"{lang.upper():<12} "
        f"{results[lang][1]:.3f} ({results[lang][1]*100:.1f}%){' '*2} "
        f"{results[lang][5]:.3f} ({results[lang][5]*100:.1f}%){' '*2} "
        f"{results[lang][10]:.3f} ({results[lang][10]*100:.1f}%)")
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
from tqdm import tqdm

# Step 1: Load the Alternative Model

print("Loading BGE-M3 model for improved Chinese retrieval")
model = SentenceTransformer('BAAI/bge-m3')
print("Model loaded successfully")


```

```python
# Step 2: Encode Articles (Same as Before)

# Encode English articles (do this once)
english_articles = [r for r in articles if r['lang'] == 'en']
chinese_articles = [r for r in articles if r['lang'] == 'zh']

english_texts = [f"{art['title']} {art['text']}" for art in english_articles]
chinese_texts = [f"{art['title']} {art['text']}" for art in chinese_articles]

english_id_to_idx = {art['id']: i for i, art in enumerate(english_articles)}
ground_truth_indices = [english_id_to_idx[art['id']] for art in chinese_articles]

print(f"Prepared {len(english_texts)} English texts")
print(f"Prepared {len(chinese_texts)} Chinese texts")
print(f"Prepared {len(ground_truth_indices)} ground truth labels")

print("\nEncoding English articles...")
english_embeddings = model.encode(
    english_texts,  # Your list of English article texts
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True
)
print(f"Encoded {len(english_embeddings)} English articles")

# Encode Chinese test articles
print("\nEncoding Chinese articles...")
chinese_embeddings = model.encode(
    chinese_texts,  # Your list of Chinese article texts
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True
)
print(f"Encoded {len(chinese_embeddings)} Chinese articles")
```

```python
# Step 3: Retrieve Top-K Candidates (Same as Before)

# Compute similarity matrix
print("\nComputing similarities and retrieving candidates")
from sklearn.metrics.pairwise import cosine_similarity

similarities = cosine_similarity(chinese_embeddings, english_embeddings)

# Get top-10 predictions for each Chinese article
predictions = []

for i in range(len(chinese_embeddings)):
    # Get indices of top-10 most similar English articles
    top_k_indices = np.argsort(similarities[i])[::-1][:10]
    predictions.append(top_k_indices.tolist())

print("Retrieved top-10 candidates for each Chinese article")
```

```python
# Evaluation based on the similarities computed previously

print("RESULTS - BGE-M3 on Chinese Articles")

for k in [1, 5, 10]:
    recall, correct, total = compute_recall_at_k(
        similarities,
        chinese_articles,
        english_articles,
        k=k
    )
    print(f"Recall@{k}: {recall:.4f} ({recall*100:.1f}%) - {correct}/{total} correct")

# Compare with Baseline
print("COMPARISON WITH BASELINE")

# Get R@1 for the new approach
new_r1, correct, total = compute_recall_at_k(
    similarities,
    chinese_articles,
    english_articles,
    k=1
)

baseline_r1 = 0.717
improvement_r1 = (new_r1 - baseline_r1) / baseline_r1 * 100

new_r2, correct, total = compute_recall_at_k(
    similarities,
    chinese_articles,
    english_articles,
    k=5
)
baseline_r2 = 0.831
improvement_r2 = (new_r2 - baseline_r2) / baseline_r2 * 100

new_r3, correct, total = compute_recall_at_k(
    similarities,
    chinese_articles,
    english_articles,
    k=10
)
baseline_r3 = 0.876
improvement_r3 = (new_r3 - baseline_r3) / baseline_r3 * 100

print(f"Baseline (LaBSE) R@1: {baseline_r1:.3f} ({baseline_r1*100:.1f}%)")
print(f"BGE-M3 R@1:           {new_r1:.3f} ({new_r1*100:.1f}%)")
print(f"Improvement:          {improvement_r1:+.1f}%")

print(f"\n\nBaseline (LaBSE) R@5: {baseline_r2:.3f} ({baseline_r2*100:.1f}%)")
print(f"BGE-M3 R@5:           {new_r2:.3f} ({new_r2*100:.1f}%)")
print(f"Improvement:          {improvement_r2:+.1f}%")

print(f"\n\nBaseline (LaBSE) R@10: {baseline_r3:.3f} ({baseline_r3*100:.1f}%)")
print(f"BGE-M3 R@10:           {new_r3:.3f} ({new_r3*100:.1f}%)")
print(f"Improvement:          {improvement_r3:+.1f}%")
```
